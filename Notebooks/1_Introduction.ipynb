## Purpose and intended audience <a id="Purpose"></a>

The book is primarily intended for undergraduate students, early-stage graduate students, engineers, and practitioners who want to approach machine learning without assuming previous programming experience. Readers with some knowledge of Python may skip or review selectively the first part, whereas readers who are new to programming are encouraged to follow the material in sequence.


## How to use this book <a id="How"></a>

Readers without previous Python experience should follow the programming chapters in order and execute the examples while reading. Modifying parameters, introducing small errors, and inspecting intermediate variables are effective ways to understand both the language and the algorithms. The exercises at the end of the chapters are intended to reinforce this active approach.

Readers who already know Python may begin with the introduction to machine learning and return to the programming chapters when a particular language feature or library is needed. Similarly, readers mainly interested in applications may initially use the library-based implementations and later study the corresponding from-scratch versions for a deeper understanding.

The numerical results of stochastic algorithms may vary slightly across computers, software versions, and random seeds. For this reason, the book emphasizes trends, methodology, and reproducible experiment design rather than exact replication of every printed value.


## Organization of the book <a id="Organization"></a>

The book is organized into two main parts followed by a collection of appendices.

The first part introduces the Python environment and the programming concepts required in the remainder of the book. It covers variables, control structures, functions, containers, strings, files, object-oriented programming, and the principal libraries for numerical computation, data manipulation, and visualization. No previous programming experience is required.

The second part is devoted to machine learning. It begins with the main learning paradigms and a brief historical perspective. Optimization and adaptive filtering are then introduced as useful foundations for iterative learning algorithms. Representative regression, classification, and clustering methods are implemented from scratch before the Scikit-learn workflow is presented.

Subsequent chapters discuss data preparation, dimensionality reduction, model evaluation, hyperparameter selection, ensemble learning, underfitting and overfitting, probability-density estimation, clustering, anomaly detection, model interpretation, imbalanced learning, and probability calibration. Complete projects combine these elements into end-to-end applications. A concise chapter on reinforcement learning introduces sequential decision-making, while the concluding chapter summarizes the main lessons and points to related paradigms.

The appendices provide compact background material on data representation, probability and stochastic processes, estimation theory, metaheuristic optimization, reproducible machine learning, and ethics and responsible machine learning. They may be read when needed or used for later reference.

The present version of the book focuses primarily on classical machine-learning methods. A future extension may introduce a separate part devoted to deep learning. The material developed here is nevertheless self-contained and provides the mathematical, computational, and methodological foundations required for that transition, including optimization, regularization, neural computation, evaluation, and reproducibility.
