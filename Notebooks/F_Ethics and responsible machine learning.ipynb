Machine learning systems increasingly influence decisions in areas such as healthcare, employment, credit, education, security, and public services. Their use can produce substantial benefits, but it can also affect privacy, equality, autonomy, safety, and access to opportunities. **Ethical** machine learning therefore requires more than high predictive accuracy: it requires careful consideration of how data are collected, how models are evaluated, how decisions are made, and who is responsible for their consequences.

Ethical questions are application dependent. A variable that is appropriate in one context may be unacceptable in another. For example, age or biological sex may be clinically relevant in a medical model but inappropriate as a basis for denying access to employment or credit. Technical performance must therefore be evaluated together with the intended purpose, the affected population, and the possible harms caused by errors.

Machine learning is not an autonomous moral agent: people and organizations choose the objectives, data, thresholds, deployment conditions, and actions taken from its outputs. Responsible practice therefore combines technical competence with awareness of social context, stakeholder participation, documentation, and continuing oversight.

The central question is not only whether a model can make a prediction, but whether that prediction should be made, how it will be used, who may be harmed, and how errors can be detected and corrected. Ethical machine learning is consequently part of good engineering rather than an optional addition to it.
