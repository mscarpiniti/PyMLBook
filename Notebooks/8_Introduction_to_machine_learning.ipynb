## Introduction <a id="Introduction"></a>

**Machine Learning** (ML) is a branch of computer science concerned with developing algorithms that learn useful patterns from data. Rather than being programmed with a complete set of rules for every possible situation, an ML system adjusts its internal parameters by using examples of the phenomenon under study. A widely used informal definition, commonly attributed to Arthur Samuel, is the following:

> Machine learning is a technique that gives computers the ability to learn a task *without* being explicitly programmed for every possible case.


## How machine learning works <a id="How_works"></a>

The fundamental *ingredients* of an ML method are therefore:
1. **Data**, which provide the examples used to train and evaluate the model.
2. A **model**, which maps the input data to an output.
3. A **cost function**, which measures the discrepancy between the model output and the desired behavior.
4. An **optimization algorithm**, which adjusts the model parameters to reduce the cost function.

## Learning paradigms <a id="Paradigms"></a>

The main learning paradigms are:
1. **Supervised learning**: the training set contains input examples together with their correct outputs, usually called *targets* or *labels*.
2. **Unsupervised learning**: the training data *do not* include target outputs.
3. **Semi-supervised learning**: the dataset contains both *labeled* and *unlabeled* examples.
4. **Reinforcement learning**: an agent interacts with an environment. At each step, the agent observes a *state*, selects an *action*, and receives a *reward*.
5. **Evolutionary learning**: a *population* of candidate solutions, also called *individuals*, is iteratively improved.
